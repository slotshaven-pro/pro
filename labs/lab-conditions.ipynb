## Sammenligninger

En betingelse kan være sand (`True`) eller falsk (`False`).

- `==` undersøger, om to værdier er ens. `=` bruges til at tildele en værdi til en variabel.
- `!=` betyder forskellig fra.
- `<` og `>` betyder mindre end og større end.
- `<=` og `>=` betyder mindre end eller lig med og større end eller lig med.

### Eksempel: Er svaret korrekt?


````python
svar = "python"
print(svar == "python")  # True
print(svar == "Python")  # False: store og små bogstaver er forskellige
````

## if og indrykning

`if` kører en blok kode, hvis betingelsen er sand.

- Linjen med `if` afsluttes med et kolon (`:`).
- Blokken efter kolon skal rykkes ind. Brug fire mellemrum pr. niveau.
- Linjer i samme blok har samme indrykning. Bland ikke tabulatorer og mellemrum.
- Når indrykningen går tilbage, er blokken slut.

### Eksempel: Hvad hører til if-blokken?

De to indrykkede linjer køres kun, når `point >= 10`. Den sidste linje køres altid.


````python
point = 12

if point >= 10:
    print("Du har vundet")
    print("Godt gået")
print("Spillet er slut")
````

## if-elif-else

`elif` betyder, at en ny betingelse undersøges, hvis de tidligere betingelser var falske. Python kører kun den første blok, hvis betingelse er sand. `else` køres, hvis ingen af betingelserne er sande.

`if`, `elif` og `else` står på samme indrykningsniveau.

### Eksempel: Beskriv temperaturen


````python
temperatur = 15

if temperatur < 0:
    print("Frost")
elif temperatur < 20:
    print("Køligt")
else:
    print("Varmt")
````

## Flere betingelser med and og or

- `and` kræver, at begge betingelser er sande.

## Indlejrede betingelser

En betingelse kan ligge inde i en anden. Den inderste betingelse undersøges kun, hvis programmet når ind i den yderste blok.

Brug fire mellemrum for første niveau og otte for næste. Indrykningen viser også, hvilket `if` et `else` hører til.

### Eksempel: Login og adgang

Det inderste `else` hører til `if er_admin`. Det yderste hører til `if logget_ind`.


````python
logget_ind = True
er_admin = False

if logget_ind:
    print("Du er logget ind")
    if er_admin:
        print("Åbn administration")
    else:
        print("Åbn elevside")
else:
    print("Log ind først")
print("Tjek afsluttet")
````
