## Iteration af strenge

Når man itererer over en streng, gennemløber man tegnene et ad gangen.

Det kan gøres med en `for`-løkke.

### Eksempel: Print hvert tegn

Her gennemløber vi strengen `hej` og printer hvert tegn på sin egen linje.

````python
tekst = "hej"

for tegn in tekst:
    print(tegn)
````

## Slicing med []

Slicing bruges til at tage en del af en streng.

Syntaksen er:

```python
streng[start:stop]
```

`start` er med, men `stop` er ikke med. Hvis vi skriver `tekst[0:4]`, får vi derfor tegnene med indeks `0`, `1`, `2` og `3`, men ikke tegnet med indeks `4`.

Man kan også skrive slicing med et tredje tal:

```python
streng[start:stop:trin]
```

`trin` bestemmer, hvor mange pladser Python hopper for hver gang. Et trin på `2` tager hvert andet tegn. Et trin på `-1` går baglæns gennem strengen.


### Eksempel: Tag de første tegn

Her tager vi de første fire tegn fra en streng.

````python
tekst = "programmering"
print(tekst[0:4])
````

### Negative indeks

Negative indeks tæller bagfra i strengen.

Det sidste tegn har indeks `-1`, det næstsidste har indeks `-2`, og sådan fortsætter det baglæns.

Det er praktisk, når man vil hente noget fra slutningen af en streng uden først at tælle alle tegnene.


````python
tekst = "programmering"

print(tekst[-1])    # sidste tegn
print(tekst[-4:])   # de sidste fire tegn
print(tekst[-7:-4]) # fra indeks -7 til -4, men -4 er ikke med
````

### Udeladte indeks

I en slice må man udelade `start` eller `stop`.

Hvis `start` udelades, starter Python fra begyndelsen af strengen. Hvis `stop` udelades, fortsætter Python helt til slutningen.

Det gør koden kortere, når man vil have begyndelsen eller slutningen af en streng.


````python
tekst = "datavidenskab"

print(tekst[:5])  # fra starten til indeks 5, men 5 er ikke med
print(tekst[4:])  # fra indeks 4 og resten af strengen
print(tekst[:])   # hele strengen
````

### Vend en streng om

Man kan vende en streng om med slicing ved at bruge trin `-1`.

Når trinnet er negativt, går Python baglæns. Med `tekst[::-1]` udelader vi både `start` og `stop`, så Python starter ved slutningen og fortsætter baglæns helt til begyndelsen.


````python
tekst = "abcdef"

print(tekst[::-1])
````

## Substrings med slicing

Man kan bruge slicing til at hente bestemte dele af en tekst. Det kræver, at man kender indeksene for den del, man vil have fat i.

### Eksempel: Hent et ord fra en sætning

Her henter vi ordet `Python` fra en længere tekst.

````python
tekst = "Jeg lærer Python"
print(tekst[10:16])
````
