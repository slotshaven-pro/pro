## for-in

En `for`-løkke gennemløber elementerne i fx en streng eller en liste ét ad gangen. Variablen efter `for` får det næste element ved hver gentagelse.

Linjen med `for ... in ...` slutter med kolon (`:`). Blokken under den skal rykkes ind; brug fire mellemrum. Når indrykningen går tilbage, er løkkens blok slut.

### Eksempel: Print hvert tegn

Den indrykkede linje køres én gang pr. tegn. Den sidste linje køres én gang efter løkken.


````python
tekst = "hej"

for tegn in tekst:
    print(tegn)
print("Færdig")
````

## Gentag med range()

`range()` giver en talfølge, som en `for`-løkke kan gennemløbe.

- `range(1, 5)` giver 1, 2, 3, 4.

Startværdien er med, men stopværdien er **ikke** med. Det tredje tal er trinnet; det er 1, hvis det udelades.

### Eksempel: Tæl fra 1 til 3


````python
for tal in range(1, 4):
    print(tal)
````

## while

En `while`-løkke gentager sin blok, så længe betingelsen er sand. Betingelsen undersøges før hver gentagelse, så blokken kan blive kørt nul gange.

Ligesom ved `for` skal linjen slutte med kolon, og blokken skal rykkes ind. Sørg for, at noget i løkken kan gøre betingelsen falsk. Ellers kan løkken fortsætte uden at stoppe.

### Eksempel: Tæl til tre

Opdateringen af `tal` ligger inde i løkken. Hvis den flyttes udenfor, bliver `tal` ved med at være 1, og løkken stopper ikke.


````python
tal = 1

while tal <= 3:
    print(tal)
    tal = tal + 1
print("Færdig")
````

## Forskellen på for-in og while

Brug typisk `for-in`, når du vil gennemløbe elementer, fx tegn i en tekst, eller gentage et kendt antal gange med `range()`.

Brug typisk `while`, når gentagelsen styres af en betingelse, fx indtil en opsparing når et mål. Antallet af gentagelser behøver ikke være kendt på forhånd.

Begge kan bruges til at tælle. En `for`-løkke henter automatisk næste værdi; i en `while`-løkke skal din kode selv sørge for, at betingelsen kan ændre sig.

### Eksempel: Spar op til et mål


````python
opsparing = 0
mål = 100
uger = 0

while opsparing < mål:
    opsparing = opsparing + 30
    uger = uger + 1
print(f"Efter {uger} uger har du {opsparing} kr.")
````

## Stop en løkke med break

`break` stopper løkken med det samme. Programmet fortsætter med koden efter løkken. Det virker i både `for` og `while`. Hvis der er løkker inde i hinanden, stopper `break` kun den inderste løkke, som det står i.

### Eksempel: Stop ved et punktum

Vi printer tegnene indtil det første punktum. Punktummet bliver ikke printet, fordi `break` står før `print(tegn)`.

Bemærk indrykningen: `break` hører til `if`-blokken, mens `print(tegn)` hører til løkken, men står uden for `if`-blokken.


````python
tekst = "Hej. Farvel."

for tegn in tekst:
    if tegn == ".":
        break
    print(tegn)
print("Løkken er slut")
````

## Spring en gentagelse over med continue

`continue` springer resten af den aktuelle gentagelse over. Løkken stopper ikke: En `for`-løkke går videre til næste element, og en `while`-løkke undersøger betingelsen igen.

### Eksempel: Spring mellemrum over

Når tegnet er et mellemrum, springes `print(tegn)` over. De øvrige tegn bliver printet.


````python
tekst = "hej med dig"

for tegn in tekst:
    if tegn == " ":
        continue
    print(tegn)
print("Færdig")
````

## continue i en while-løkke

Pas på, at `continue` ikke springer opdateringen af løkkens variabel over. Ellers kan løkken blive ved uden at stoppe.

### Eksempel: Spring tallet 3 over

Her opdateres `tal` før `continue`, så løkken også kommer videre, når tallet er 3.


````python
tal = 0

while tal < 5:
    tal = tal + 1
    if tal == 3:
        continue
    print(tal)
print("Færdig")
````
