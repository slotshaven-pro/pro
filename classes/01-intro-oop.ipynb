## Del 1 – Gennemgang

### 1.1 Alt i Python er objekter

Du har allerede brugt objekter rigtig meget – uden at vide det. Et tal, en tekst, en liste og en dictionary er alle **objekter**.

Et objekt er en "ting" i hukommelsen, som har:

- **data** (fx teksten `"hej"` eller tallet `42`)
- **metoder** – funktioner der hører til objektet (fx `.upper()` eller `.append()`)

Hvert objekt er lavet ud fra en **klasse**. Klassen bestemmer, hvilken *type* objektet har, og hvilke metoder det kan bruge. Funktionen `type()` fortæller, hvilken klasse et objekt er lavet ud fra.

````python
print(type(42))
print(type(3.14))
print(type("hej"))
print(type([1, 2, 3]))
print(type({"navn": "Ida"}))
print(type(True))
````

Læg mærke til ordet `class` i resultatet: `42` er et objekt af klassen `int`, og `"hej"` er et objekt af klassen `str`.

Når du skriver `.upper()` efter en tekst, kalder du en **metode**, som klassen `str` har defineret:

````python
tekst = "hej verden"
print(tekst.upper())        # metode fra klassen str

tal = [3, 1, 2]
tal.append(10)              # metode fra klassen list
print(tal)
````

Selv funktioner er objekter i Python:

````python
print(type(print))
print(type(len))
````

**Pointen:** Python har en masse indbyggede klasser (`int`, `str`, `list`, `dict`, ...). I objektorienteret programmering lærer vi at lave **vores egne klasser**.

### 1.2 Klasse og instans

- En **klasse** er en *skabelon* eller en *opskrift*. Den beskriver, hvilke data og hvilke metoder objekterne skal have.
- En **instans** er et konkret objekt, der er lavet ud fra klassen. Ordet *objekt* og *instans* betyder stort set det samme.

**Sammenligning:** En tegning af et hus er klassen. Husene, der bliver bygget ud fra tegningen, er instanserne. Man kan bygge mange huse ud fra én tegning – og hvert hus kan have sin egen farve.

| Klasse | Instanser |
|---|---|
| `str` | `"hej"`, `"Python"`, `""` |
| `list` | `[1, 2]`, `["a", "b", "c"]` |
| `Hund` (vores egen) | Fido, Bella, Rex |

### 1.3 Den tomme klasse

En klasse laves med nøgleordet `class`. Klassenavne skrives med stort begyndelsesbogstav.

````python
class Hund:
    pass    # pass betyder "gør ingenting" – klassen er tom
````

Nu kan vi lave **instanser** af klassen ved at kalde klassen som en funktion: `Hund()`.

````python
fido = Hund()
bella = Hund()

print(type(fido))
print(fido)
print(bella)
print(fido is bella)    # er det det samme objekt?
````

`fido` og `bella` er to **forskellige** objekter (de har forskellige adresser i hukommelsen, fx `0x10a3...`), men de er lavet ud fra den samme klasse.

### 1.4 En attribut i klassen

En **attribut** (også kaldet en egenskab) er en variabel, der hører til et objekt. Her giver vi klassen en attribut `lyd`:

````python
class Hund:
    lyd = "Vov"

fido = Hund()
print(fido.lyd)     # attributter tilgås med punktum
````

### 1.5 En metode

En **metode** er en funktion, der står inde i klassen. Den første parameter hedder altid `self`.

`self` er **det objekt, metoden bliver kaldt på**. Når vi skriver `fido.sig_noget()`, sender Python automatisk `fido` med som `self`.

````python
class Hund:
    lyd = "Vov"

    def sig_noget(self):
        print("Hunden siger:", self.lyd)

fido = Hund()
fido.sig_noget()    # Python laver det om til: Hund.sig_noget(fido)
````

**Problemet:** Alle hunde hedder ingenting, og vi kan ikke give dem forskellige værdier, når vi laver dem. Det løser vi med en **konstruktør**.

### 1.6 Konstruktøren `__init__`

Konstruktøren er en særlig metode, der hedder `__init__` (to understreger før og efter). Den bliver kørt **automatisk**, hver gang en ny instans bliver lavet.

I konstruktøren gemmer vi værdier på objektet med `self.navn = navn`. På den måde får **hver instans sine egne værdier**.

````python
class Hund:
    def __init__(self, navn):
        self.navn = navn            # gem navnet på DETTE objekt

    def sig_noget(self):
        print(self.navn, "siger Vov")

fido = Hund("Fido")      # "Fido" sendes til parameteren navn i __init__
bella = Hund("Bella")

fido.sig_noget()
bella.sig_noget()
print(fido.navn)
````

Læg mærke til:

- Vi kalder **ikke** `__init__` selv – det sker, når vi skriver `Hund("Fido")`.
- `self.navn` er objektets attribut. `navn` (uden `self`) er bare parameteren, som forsvinder, når konstruktøren er færdig.
- `fido` og `bella` er lavet ud fra samme klasse, men har **forskellige** værdier.

### Begreber – opsamling

| Begreb | Forklaring | Eksempel |
|---|---|---|
| Klasse | Skabelon for objekter | `class Hund:` |
| Instans / objekt | Konkret objekt lavet ud fra en klasse | `fido = Hund("Fido")` |
| Attribut | Variabel der hører til objektet | `self.navn` |
| Metode | Funktion der hører til objektet | `def sig_noget(self):` |
| `self` | Det objekt, metoden kaldes på | `self.navn` |
| Konstruktør | Metode der kører, når objektet laves | `def __init__(self, navn):` |
