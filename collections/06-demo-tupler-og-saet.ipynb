# Tupler og sæt i Python

Denne notebook viser de to sidste kollektioner: **tupler** og **sæt**.

**Fremgangsmåde:** Læs forklaringen, kør eksemplet – og ret gerne i koden for at se, hvad der sker.

| | Tupel `( )` | Sæt `{ }` |
|---|---|---|
| Ordnet | ja | **nej** |
| Kan ændres | **nej** | ja |
| Dubletter | ja | **nej** |
| Adgang | indeks | kun `in` |

# Del 1: Tupler

### Opret en tupel

En tupel oprettes med runde parenteser `( )`. Den ligner en liste, men kan **ikke** ændres.

```python
koordinat = (55.68, 12.57)
dato = (2026, 9, 28)
print(koordinat)
print(type(dato))
```

### Tupel med ét element

**Bemærk:** En tupel med ét element skal have et **komma**. Uden komma er parentesen bare en almindelig parentes.

```python
a = (5)
b = (5,)
print(type(a))   # int
print(type(b))   # tuple
```

### Indeksering og slicing

Tupler er ordnede, så man kan bruge indeks og slicing – præcis som med lister.

```python
ugedage = ("man", "tirs", "ons", "tors", "fre")
print(ugedage[0])
print(ugedage[-1])
print(ugedage[1:3])
print(len(ugedage))
```

### Tupler kan ikke ændres

Man kan ikke ændre, tilføje eller fjerne elementer. Tupler har derfor ingen `append()`, `remove()` eller `sort()`.

```python
punkt = (3, 4)
try:
    punkt[0] = 99
except TypeError as fejl:
    print("Fejl:", fejl)
```

Vil man "ændre" en tupel, må man lave en **ny**:

```python
punkt = (3, 4)
punkt = (99, punkt[1])
print(punkt)
```

### Udpakning

Værdierne i en tupel kan pakkes ud i hver sin variabel. Antallet af variabler skal passe med antallet af elementer.

```python
punkt = (3, 4)
x, y = punkt
print(f"x = {x}, y = {y}")

år, måned, dag = (2026, 9, 28)
print(dag, måned, år)
```

### Funktioner der returnerer flere værdier

Når en funktion returnerer flere værdier adskilt af komma, er det i virkeligheden en tupel.

```python
def min_og_max(tal):
    return min(tal), max(tal)

resultat = min_og_max([5, 1, 9, 2])
print(resultat, type(resultat))

laveste, højeste = min_og_max([5, 1, 9, 2])
print(laveste, højeste)
```

### Metoderne `count()` og `index()`

Tupler har kun to metoder – begge ændrer ingenting.

```python
terninger = (6, 2, 6, 3, 6)
print("Antal seksere:", terninger.count(6))
print("Første 3'er ved indeks:", terninger.index(3))
```

### Gennemløb

Tupler kan gennemløbes med `for` ligesom lister. Ofte ser man en **liste af tupler**, hvor hver tupel er en "række".

```python
aktier = [("Novo", 780.5), ("Maersk", 11200.0), ("Carlsberg", 850.2)]
for navn, kurs in aktier:
    print(f"{navn}: {kurs} kr.")
```

# Del 2: Sæt

### Opret et sæt

Et sæt oprettes med krøllede parenteser `{ }` – **uden** kolon.

**Bemærk:** Et **tomt** sæt skal laves med `set()`. `{}` er et tomt dictionary!

```python
farver = {"rød", "grøn", "blå"}
print(farver)
print(type(farver))

tom = set()
print(type(tom))       # set
print(type({}))        # dict!
```

### Sæt fjerner dubletter

Et sæt kan kun indeholde hver værdi én gang. Det er en nem måde at fjerne dubletter fra en liste på.

```python
karakterer = [7, 10, 7, 12, 10, 7]
unikke = set(karakterer)
print(unikke)
print("Antal forskellige karakterer:", len(unikke))
```

### Sæt er ikke ordnede

Et sæt har ingen fast rækkefølge og **intet indeks**. Man kan derfor ikke skrive `farver[0]`.

Man kan kun spørge, om noget **er med** – med `in`.

```python
farver = {"rød", "grøn", "blå"}
print("rød" in farver)
print("gul" in farver)

try:
    print(farver[0])
except TypeError as fejl:
    print("Fejl:", fejl)
```

### Tilføj og fjern: `add()`, `remove()` og `discard()`

- `add()` tilføjer et element (sker ingenting, hvis det allerede er der).
- `remove()` fjerner et element – giver `KeyError`, hvis det ikke findes.
- `discard()` fjerner et element – giver **ingen** fejl, hvis det ikke findes.

```python
s = {"rød", "grøn"}
s.add("blå")
s.add("rød")          # findes allerede – ingen ændring
s.discard("gul")      # findes ikke – ingen fejl
s.remove("grøn")
print(s)
```

### Mængdeoperationer

Sæt kan det samme som mængder i matematik:

| Operation | Tegn | Betydning |
|---|---|---|
| foreningsmængde | `A \| B` | alle elementer fra A og B |
| fællesmængde | `A & B` | elementer, der er i både A og B |
| differensmængde | `A - B` | elementer i A, som ikke er i B |

```python
fodbold = {"Ida", "Jon", "Kai", "Lea"}
håndbold = {"Kai", "Lea", "Mia"}

print("Går til noget:", fodbold | håndbold)
print("Går til begge:", fodbold & håndbold)
print("Kun fodbold:", fodbold - håndbold)
```

### Gennemløb

Et sæt kan gennemløbes med `for` – men rækkefølgen er ikke til at forudsige. Skal man have en fast rækkefølge, kan man bruge `sorted()`, som returnerer en sorteret **liste**.

```python
navne = {"Kai", "Ida", "Lea", "Jon"}
for navn in sorted(navne):
    print(navn)
```

# Del 3: Konvertering mellem typerne

Man kan konvertere mellem kollektionerne med typens navn: `list()`, `tuple()`, `set()`, `dict()`.

```python
tal = [3, 1, 3, 2]
print(tuple(tal))       # (3, 1, 3, 2)
print(set(tal))         # {1, 2, 3}
print(sorted(set(tal))) # [1, 2, 3] – unikke og sorteret, som liste
```

```python
# Et dictionary kan laves ud fra en liste af tupler
par = [("Ida", 12), ("Jon", 7)]
karakterer = dict(par)
print(karakterer)

# ... og items() giver par tilbage som tupler
print(list(karakterer.items()))
```

## Ekstra: Set comprehension og delmængder

_Denne del er for dem, der vil vide mere._

```python
ordliste = ["hej", "hejsa", "hej", "hej", "python"]
længder = {len(o) for o in ordliste}     # set comprehension
print(længder)

A = {1, 2}
B = {1, 2, 3, 4}
print(A <= B)    # er A en delmængde af B?
print(A ^ B)     # symmetrisk differens: i A eller B, men ikke i begge
```
