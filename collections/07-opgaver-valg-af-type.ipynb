### Opgave 9 – Fra liste til sæt (mellem)

Koden finder de elever, der går til både fodbold og håndbold. Omskriv den, så den bruger **sæt** og mængdeoperationer. Hvor mange linjer kan du spare?

````python
fodbold = ["Ida", "Jon", "Kai", "Lea"]
håndbold = ["Kai", "Lea", "Mia"]

begge = []
for navn in fodbold:
    if navn in håndbold:
        begge.append(navn)
print(begge)
````
