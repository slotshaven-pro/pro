# Dictionaries i Python

Et **dictionary** (på dansk også _opslagsliste_ eller _ordbog_) er en samling af **nøgle–værdi-par**. Man slår en værdi op via dens nøgle – ligesom man i en ordbog slår et ord op for at finde forklaringen.

**Fremgangsmåde:** Læs forklaringen, kør eksemplet – og ret gerne i koden for at se, hvad der sker.

**Egenskaber for dictionaries:** ordnet efter indsættelse · kan ændres · nøgler er unikke · adgang via nøgle

## Del 1: Oprettelse og opslag

### Opret et dictionary

Et dictionary oprettes med krøllede parenteser `{ }`. Hvert par skrives `nøgle: værdi`, og parrene adskilles med komma.

**Syntaks**
`navn = {nøgle1: værdi1, nøgle2: værdi2}`

**Bemærk:** Det er **kolon** mellem nøgle og værdi, der gør det til et dictionary. `{"a", "b"}` uden kolon er et **sæt**.

```python
postnumre = {"København": 1000, "Aarhus": 8000, "Odense": 5000}
print(postnumre)
print(type(postnumre))
```

### Det tomme dictionary

`{}` er et tomt dictionary. Det kan fyldes op senere.

```python
lager = {}
print(len(lager))    # 0
lager["mælk"] = 12
print(lager)
```

### Læs en værdi via nøglen

**Syntaks**
`værdi = d[nøgle]`

**Bemærk:** Hvis nøglen ikke findes, får man en `KeyError`.

```python
telefoner = {"Anna": "22 11 00 33", "Bo": "40 12 34 56"}
print(telefoner["Anna"])
# print(telefoner["Carl"])   # fjern # og se fejlen: KeyError
```

### Skriv en værdi via nøglen

Man bruger samme syntaks til at **ændre** en eksisterende værdi og til at **tilføje** et nyt par.

**Syntaks**
`d[nøgle] = værdi`

- Findes nøglen i forvejen, bliver værdien overskrevet.
- Findes nøglen ikke, bliver parret tilføjet.

```python
lager = {"rugbrød": 10}
lager["rugbrød"] = lager["rugbrød"] + 5   # ændring: 10 → 15
lager["smør"] = 7                         # ny nøgle
print(lager)
```

### Nøgler er unikke

Et dictionary kan ikke have den samme nøgle to gange. Hvis man skriver den samme nøgle igen, overskrives den gamle værdi – **uden advarsel**.

Værdierne må derimod godt være ens.

```python
menu = {"smørrebrød": 55, "smørrebrød": 60, "frikadeller": 95}
print(menu)          # kun den sidste smørrebrød-pris er gemt

alder = {"Ida": 17, "Jon": 17}   # to forskellige nøgler med samme værdi – OK
print(alder)
```

### Test om en nøgle findes med `in`

**Syntaks**
`nøgle in d`

**Bemærk:** `in` tester kun **nøgler** – ikke værdier.

```python
klasse = {"Freja": 9, "Noah": 8, "Alma": 9}   # navn: klassetrin
print("Freja" in klasse)   # True
print(9 in klasse)         # False – 9 er en værdi, ikke en nøgle

navn = "Carl"
if navn in klasse:
    print(klasse[navn])
else:
    print(navn, "findes ikke")
```

### Sikker læsning med `get()`

Metoden `get()` giver aldrig en fejl. Findes nøglen ikke, returneres `None` – eller den **standardværdi** (default), man selv angiver.

**Syntaks**
`værdi = d.get(nøgle, standardværdi)`

```python
lager = {"mælk": 12}
print(lager.get("mælk"))          # 12
print(lager.get("yoghurt"))       # None
print(lager.get("yoghurt", 0))    # 0
```

## Del 2: Gennemløb

### Gennemløb med `for`

Når man gennemløber et dictionary med `for`, får man **nøglerne**.

**Syntaks**
`for nøgle in d:`

```python
indbyggere = {"Hvidovre": 54000, "Gladsaxe": 70000, "Lyngby": 56000}
for kommune in indbyggere:
    print(kommune, indbyggere[kommune])
```

### Hent både nøgle og værdi med `items()`

Med `items()` får man både nøgle og værdi i hvert gennemløb. Det er den mest brugte måde at gennemløbe et dictionary på.

**Syntaks**
`for nøgle, værdi in d.items():`

**Bemærk:** Hvert par er en **tupel** `(nøgle, værdi)`, som bliver pakket ud i to variabler.

```python
skema = {"historie": "11:00", "biologi": "12:00"}
for fag, tid in skema.items():
    print(f"{fag}: {tid}")
```

### Alle nøgler med `keys()` og alle værdier med `values()`

```python
bon = {"kaffe": 30, "bolle": 20, "juice": 25}
print(list(bon.keys()))      # ['kaffe', 'bolle', 'juice']
print(list(bon.values()))    # [30, 20, 25]
print("I alt:", sum(bon.values()))
```

### Gennemløb i sorteret rækkefølge

`sorted(d)` giver en sorteret **liste** af nøglerne.

```python
data = {"b": 2, "c": 3, "a": 1}
for nøgle in sorted(data):
    print(nøgle, data[nøgle])
```

## Del 3: Fjern elementer

### Fjern med `del`

**Syntaks**
`del d[nøgle]`

**Bemærk:** `del` er et **keyword** – ikke en funktion eller metode.

```python
kurv = {"banan": 3, "æble": 5, "pære": 2}
del kurv["banan"]
print(kurv)
```

### Fjern og returnér værdien med `pop()`

`pop()` fjerner parret og **returnerer** værdien, så man kan gemme den.

**Syntaks**
`værdi = d.pop(nøgle, standardværdi)`

**Bemærk:** Angiv en standardværdi for at undgå `KeyError`, hvis nøglen ikke findes.

```python
kurv = {"banan": 3, "æble": 5, "pære": 2}
antal = kurv.pop("æble")
print("Fjernede æbler:", antal)
print("Kurven:", kurv)
print(kurv.pop("kiwi", 0))   # kiwi findes ikke → 0
```

## Del 4: Indlejrede data

### Værdier kan være af alle typer

Værdierne kan være tal, tekst, lister – og andre dictionaries.

**Bemærk:** Nøgler skal være **uforanderlige** (fx tekst, tal eller tupler). En liste kan ikke bruges som nøgle.

```python
rejsekort = {
    "ejer": "Mads",
    "saldo": 123.50,
    "zoner": [1, 2],
    "aktiv": True,
}
print(rejsekort["zoner"])
rejsekort["zoner"].append(3)
print(rejsekort)
```

### Dictionaries i dictionaries

Man læser en værdi dybt inde med "dobbelt nøgle": `d["Ida"]["klasse"]`.

```python
register = {
    "Ida": {"klasse": "2x", "email": "ida@skole.dk"},
    "Malthe": {"klasse": "2y", "email": "malthe@skole.dk"},
}
print(register["Ida"]["klasse"])
register["Ida"]["email"] = "ida@example.dk"
print(register["Ida"])
```

### Liste af dictionaries

En **liste af dictionaries** er en meget almindelig måde at gemme "rækker" af data på – fx når man henter data fra en database eller et API. Hver dictionary er én række, og nøglerne er kolonnenavnene.

```python
elever = [
    {"navn": "Ida", "klasse": "2x", "karakter": 12},
    {"navn": "Jon", "klasse": "2y", "karakter": 7},
    {"navn": "Kai", "klasse": "2x", "karakter": 10},
]

for elev in elever:
    print(f"{elev['navn']} går i {elev['klasse']} og fik {elev['karakter']}")
```

## Del 5: Byg et dictionary med en løkke

### Optælling

Et typisk mønster: gennemløb en liste og tæl, hvor mange gange hver værdi optræder.

```python
stemmer = ["pizza", "burger", "pizza", "sushi", "pizza", "burger"]
optælling = {}
for ret in stemmer:
    if ret in optælling:
        optælling[ret] = optælling[ret] + 1
    else:
        optælling[ret] = 1
print(optælling)
```

Samme optælling kan skrives kortere med `get()`:

```python
optælling = {}
for ret in stemmer:
    optælling[ret] = optælling.get(ret, 0) + 1
print(optælling)
```

### Dict comprehension

Ligesom list comprehension kan man bygge et dictionary i én linje.

**Syntaks**
`{nøgle: værdi for element in liste}`

```python
kvadrater = {n: n * n for n in range(1, 6)}
print(kvadrater)

navne = ["Ida", "Jon", "Kai"]
længder = {navn: len(navn) for navn in navne}
print(længder)
```

## Ekstra: Flere metoder

_Denne del er for dem, der vil vide mere._

### Flet to dictionaries med `update()`

`update()` tilføjer alle par fra et andet dictionary. Findes en nøgle i begge, vinder den nye værdi.

```python
a = {"x": 1, "y": 2}
b = {"y": 20, "z": 3}
a.update(b)
print(a)
```

### `setdefault()`

Returnerer værdien for nøglen. Hvis nøglen ikke findes, indsættes den først med standardværdien.

```python
grupper = {}
grupper.setdefault("2x", []).append("Ida")
grupper.setdefault("2x", []).append("Kai")
grupper.setdefault("2y", []).append("Jon")
print(grupper)
```

### Tupler som nøgler

Fordi tupler er uforanderlige, kan de bruges som nøgler – fx til koordinater.

```python
kort = {}
kort[(2, 3)] = "træ"
kort[(0, 0)] = "hus"
print(kort[(2, 3)])
```
