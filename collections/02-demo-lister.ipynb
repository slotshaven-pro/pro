# Lister i Python

Her følger en række eksempler på, hvordan man arbejder med lister.

**Fremgangsmåde:** Læs forklaringen, kør eksemplet – og ret gerne i koden for at se, hvad der sker.

Læs [01-teori-kollektioner.md](01-teori-kollektioner.md) først, hvis du ikke allerede har gjort det.

**Egenskaber for lister:** ordnet · kan ændres · tillader dubletter · adgang via indeks

## Del 1: Oprettelse og egenskaber

### Opret en liste

En liste oprettes med kantede parenteser `[ ]`. Elementerne adskilles med komma.

**Syntaks**
`navn = [element1, element2, element3]`

```python
byer = ["Aarhus", "Odense", "Aalborg"]
print(byer)
print(type(byer))
```

### Den tomme liste

En tom liste har 0 elementer og kan fyldes op senere. Funktionen `len()` giver antallet af elementer.

```python
tom = []
print(len(tom))      # 0
tom.append("cykel")
print(tom)
print(len(tom))      # 1
```

### Lister er ordnede

Elementerne ligger i den rækkefølge, de blev sat ind i. Rækkefølgen ændrer sig ikke af sig selv.

```python
dage = ["mandag", "tirsdag", "onsdag"]
print(dage)
print(dage)          # samme rækkefølge hver gang
```

### Lister tillader dubletter

Den samme værdi kan stå flere gange. Metoden `count()` tæller, hvor mange gange en værdi optræder.

```python
byer = ["Aarhus", "Odense", "Aarhus", "Aalborg"]
print(byer)
print(byer.count("Aarhus"))   # 2
```

### Elementer kan være af alle datatyper

En liste kan blande tal, tekst, sandhedsværdier – og endda andre lister eller dictionaries.

**Bemærk:** I praksis er det mest almindeligt, at alle elementer i en liste har samme type.

```python
blandet = [42, "kanelsnegl", True, 3.14, ["a", "b"]]
for element in blandet:
    print(type(element), element)
```

### Opret en liste med `list()`

Typekonstruktøren `list()` kan lave en liste ud fra andre ting, man kan gennemløbe – fx en streng.

```python
tekst = "Smørrebrød"
bogstaver = list(tekst)
print(bogstaver)
```

### Opret en liste med `range()`

`range(start, stop)` giver tallene fra `start` op til – men **ikke med** – `stop`.

```python
husnumre = list(range(1, 11))   # 1..10
print(husnumre)
```

## Del 2: Adgang til elementer

### Læs og skriv med indeks

Hvert element har et **indeks** (en position). Det første element har indeks **0**.

**Syntaks**
`x = liste[indeks]` (læs)
`liste[indeks] = x` (skriv)

**Bemærk:** Negative indeks tæller bagfra: `liste[-1]` er det sidste element.

```python
hold = ["Brøndby", "FC København", "AGF", "AaB"]
print(hold[0])       # første
print(hold[-1])      # sidste
hold[2] = "Randers"  # lister kan ændres (mutable)
print(hold)
```

### IndexError

Hvis man bruger et indeks, der ikke findes, får man en fejl. Det sidste gyldige indeks er altid `len(liste) - 1`.

```python
hold = ["Brøndby", "FC København", "AGF"]
print(len(hold))     # 3 – så sidste indeks er 2
# print(hold[3])     # fjern # og se fejlen: IndexError
```

### Udsnit med slicing `[start:stop:step]`

Med **slicing** kan man tage et udsnit af en liste. Resultatet er en **ny** liste.

**Syntaks**
`liste[start:stop:step]`

- `start` – første indeks, der er med (standard: 0)
- `stop` – første indeks, der **ikke** er med (standard: listens længde)
- `step` – hvor mange der springes (standard: 1)

```python
måneder = ["jan", "feb", "mar", "apr", "maj", "jun", "jul", "aug", "sep", "okt", "nov", "dec"]
print(måneder[:3])      # de tre første
print(måneder[3:6])     # indeks 3, 4 og 5
print(måneder[-2:])     # de to sidste
print(måneder[::2])     # hver anden
print(måneder[::-1])    # baglæns
```

### Test om et element er i listen med `in`

`in` giver `True` eller `False`. Det er et **udtryk** med en sandhedsværdi, så det kan bruges direkte i en `if`.

```python
stationer = ["Nørreport", "Østerport", "Høje Taastrup"]
print("Nørreport" in stationer)   # True
print("Farum" in stationer)       # False

if "Farum" not in stationer:
    print("Farum er ikke på listen")
```

### Find positionen med `index()`

`index()` returnerer indekset for den første forekomst af en værdi.

**Bemærk:** Hvis værdien ikke findes, giver det en `ValueError`. Test derfor gerne med `in` først.

```python
stationer = ["Nørreport", "Østerport", "Høje Taastrup"]
print(stationer.index("Østerport"))   # 1
```

## Del 3: Gennemløb

### Gennemløb med `for`

Lister kan gennemløbes med en `for`-løkke – ét element ad gangen.

```python
kommuner = ["København", "Frederiksberg", "Aarhus", "Odense"]
for kommune in kommuner:
    print(f"Kommune: {kommune}")
```

### Gennemløb med indeks: `enumerate()`

Funktionen `enumerate()` giver både indeks og element. Med `start=1` begynder tællingen ved 1.

```python
navne = ["Ida", "Jon", "Kai"]
for nr, navn in enumerate(navne, start=1):
    print(nr, navn)
```

### Gennemløb af to lister på én gang: `zip()`

Funktionen `zip()` parrer elementerne fra to lister: første med første, anden med anden osv.

```python
navne = ["Ida", "Jon", "Kai"]
point = [12, 10, 15]
for navn, p in zip(navne, point):
    print(f"{navn}: {p} point")
```

### Beregninger på lister: `sum()`, `min()`, `max()`

Indbyggede funktioner, der virker på lister af tal.

```python
temperaturer = [14.5, 17.0, 12.3, 19.8, 16.1]
print("Højeste:", max(temperaturer))
print("Laveste:", min(temperaturer))
print("Gennemsnit:", sum(temperaturer) / len(temperaturer))
```

## Del 4: Ændring af lister (metoder)

Lister har en række **metoder**, som man kalder med _dot-notation_: `liste.metode()`.

**Bemærk:** De fleste af metoderne ændrer listen **direkte** (in-place) og returnerer `None`.

### Tilføj ét element bagerst med `append()`

```python
retter = ["frikadeller", "flæskesteg"]
retter.append("stegt flæsk")
print(retter)
```

### Tilføj flere elementer med `extend()`

`extend()` tilføjer alle elementer fra en anden liste.

**Bemærk:** Sammenlign med `append()`: `append([1, 2])` tilføjer hele listen som **ét** element.

```python
nordjylland = ["Skagen", "Hirtshals"]
flere = ["Frederikshavn", "Aalborg"]
nordjylland.extend(flere)
print(nordjylland)

a = [1, 2]
a.append([3, 4])
print(a)             # [1, 2, [3, 4]] – en liste inde i listen
```

### Sæt ind på en bestemt plads med `insert()`

**Syntaks**
`liste.insert(indeks, element)`

```python
s_tog = ["Klampenborg", "Hellerup", "Nørreport"]
s_tog.insert(2, "Østerport")
print(s_tog)
```

### Fjern første forekomst med `remove()`

**Bemærk:** Giver en `ValueError`, hvis værdien ikke findes.

```python
byer = ["Vejle", "Horsens", "Kolding", "Vejle"]
byer.remove("Vejle")     # kun den første Vejle fjernes
print(byer)
```

### Fjern og returnér et element med `pop()`

`pop()` fjerner det sidste element, `pop(i)` fjerner elementet på indeks `i`. I begge tilfælde **returneres** elementet, så man kan gemme det i en variabel.

```python
kø = ["kunde1", "kunde2", "kunde3"]
sidste = kø.pop()        # fjerner det sidste
første = kø.pop(0)       # fjerner indeks 0
print(sidste, første)
print(kø)
```

### Slet med `del` og tøm med `clear()`

`del` er et **keyword** (ikke en metode). Det kan slette ét element eller et udsnit.

```python
dyr = ["kat", "hund", "fugl", "fisk"]
del dyr[0]
print(dyr)
dyr.clear()
print(dyr)       # []
```

### Sortering: `sort()` og `sorted()`

Der er to måder at sortere på, og de virker forskelligt:

- `liste.sort()` er en **metode**, der ændrer listen selv (in-place).
- `sorted(liste)` er en **funktion**, der returnerer en **ny** sorteret liste. Den oprindelige er uændret.

Begge kan tage `reverse=True` for at sortere faldende.

```python
klubber = ["Silkeborg", "Lyngby", "Midtjylland", "OB"]
sorterede = sorted(klubber)     # ny liste
print(sorterede)
print(klubber)                  # uændret

klubber.sort(reverse=True)      # ændrer selve listen
print(klubber)
```

### Vend en liste med `reverse()`

```python
øer = ["Bornholm", "Fyn", "Langeland", "Als"]
øer.reverse()
print(øer)
```

## Del 5: Lister i lister

### En matrix som liste af lister

En tabel (matrix) kan laves som en liste, hvor hvert element er en række (en liste).

Man tilgår et element med **dobbelt indeks**: `matrix[række][kolonne]`.

```python
matrix = [
    [1, 2, 3],   # række 0
    [4, 5, 6],   # række 1
    [7, 8, 9],   # række 2
]
print(matrix[0])       # hele række 0
print(matrix[1][2])    # række 1, kolonne 2 → 6

for række in matrix:
    print(række)
```

## Del 6: List comprehension

List comprehension er en kort måde at bygge en ny liste ud fra en anden på – i én linje.

**Syntaks**
`[udtryk for element in liste]`
`[udtryk for element in liste if betingelse]`

Sammenlign de to celler nedenfor – de gør præcis det samme.

```python
# Med en almindelig for-løkke
kvadrater = []
for n in range(10):
    kvadrater.append(n * n)
print(kvadrater)
```

```python
# Med list comprehension
kvadrater = [n * n for n in range(10)]
print(kvadrater)
```

```python
# Med betingelse: kun lige tal
lige = [n for n in range(1, 21) if n % 2 == 0]
print(lige)
```

## Ekstra: Kopi eller samme liste?

_Denne del er for dem, der vil vide mere._

Når man skriver `b = a`, laver man **ikke** en kopi. Begge variabler peger på **den samme** liste.

```python
a = [1, 2, 3]
b = a            # samme liste – ikke en kopi
b.append(4)
print(a)         # [1, 2, 3, 4] – a er også ændret!
```

Vil man have en kopi, bruger man `copy()` (eller `a[:]` eller `list(a)`).

```python
frugter = ["æble", "pære", "blomme"]
kopi = frugter.copy()
frugter.append("banan")
print("original:", frugter)
print("kopi:    ", kopi)
```

**Overfladisk kopi:** `copy()` kopierer kun det yderste lag. Hvis listen indeholder andre lister, deles de indre lister stadig mellem original og kopi.

```python
original = [["rugbrød"], ["smør"]]
kopi = original.copy()
original[0].append("kerner")   # ændrer en indre liste
print("original:", original)
print("kopi:    ", kopi)        # den indre liste er også ændret her!
```

## Ekstra: Ændring med slicing

Man kan også erstatte, indsætte og fjerne flere elementer på én gang med slicing.

```python
tal = [1, 2, 3, 4, 5]
tal[1:3] = [20, 30, 40]   # erstat 2 og 3 med tre nye elementer
print(tal)
tal[0:2] = []             # fjern de to første
print(tal)
```
